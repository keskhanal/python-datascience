## Boolean indexing / filtering

````python
import numpy as np

arr = np.array([1,2,3,43,4,5,6])
print(arr > 5)
````

````python
print(arr[arr > 5])
````

````python
arr = np.array([10, 25, 30, 45, 50, 65])

print(arr[(arr < 30) & (arr < 60)])
````

## np.where()

````python
marks = np.array([35, 67, 42, 89, 28])

result = np.where(marks >= 40, "Pass", "Fail")
print(result)
````

````python
salary = np.array([25000, 40000, 55000, 30000])

new_salary = np.where(salary <= 40000, salary*1.2, salary*1.1)
print(new_salary)
````

## sorting

````python
arr = np.array([50, 20, 80, 10, 40])

# asending
print(f"ascnding: {np.sort(arr)}")

# descending
print(f"descending: {np.sort(arr)[::-1]}")
````

````python
# argsort() - returns index instead of values
print(np.argsort(arr))
````

## unique values

````python
cities = np.array([
    "Kathmandu",
    "Pokhara",
    "Kathmandu",
    "Chitwan",
    "Pokhara"
])

print(np.unique(cities))
````

````python
cities = ["Kathmandu", "Pokhara", "Kathmandu", "Chitwan", "Pokhara"]
unique_cities = list(set(cities))
print(unique_cities)
````

````python
# count occurances
values, count = np.unique(cities, return_counts=True)
print(values)
print(count)
````

## missing value - NaN

````python
data = np.array([10, 20, np.nan, 40, 50])

print(data)
````

````python
print(np.mean(data))

new_data = np.where(np.isnan(data), 0.0, data)
new_data
````

````python
# detect missing values
print(np.isnan(data))
````

````python
# removing nan values
print(data[~np.isnan(data)])
````

## student performance analaysis
```code
marks = np.array([
    [65, 70, 75],
    [80, 85, 90],
    [45, 55, 60],
    [90, 95, 92],
    [35, 40, 50],
    [72, 68, 75],
    [88, 82, 84],
    [55, 60, 58]
])
```
columns: Python | Mathematics | Statistics

### calculate
1. Average mark of every student
2. Average mark of every subject
3. Highest mark in the dataset
4. Lowest mark
5. Student with highest average
6. Students whose Python mark is above 70
7. Number of individual marks above 80
8. Students with an overall average below 50
9. Sort students based on overall average
10. Assign a result using:
```code 
Average >= 80     → Distinction
Average >= 60     → First Division
Average >= 40     → Pass
Average < 40      → Fail
```

````python
# 10 assign results
marks = np.array([
    [65, 70, 75],
    [80, 85, 90],
    [45, 55, 60],
    [90, 95, 92],
    [35, 40, 50],
    [72, 68, 75],
    [88, 82, 84],
    [55, 60, 58],
    [10, 20, 30],
])

# average mark of every subject
sub_avg = np.mean(marks, axis=0)
print(sub_avg)

# student wise
std_avg = np.mean(marks, axis=1)
print(std_avg)

results = np.select(
    [
        std_avg >= 80,
        std_avg >= 60,
        std_avg >= 40
    ],
    [
        'Distinction',
        "First Division",
        "Pass"
    ],
    default="Fail"
)

print(results)
````
