# why numpy array instead of python list

```python
# using list
numbers = [1, 2, 3, 4]

# This doesn't multiply individual elements
print(numbers * 3)

new_numbers = [number * 3 for number in numbers]
print(new_numbers)
```

```python
numbers = np.array([1,2,3,4])
print(numbers * 2)
```

## slicing in numpy

arr[start:stop:step]

```python
arr = np.array([1,2,3,4,5,6,7,9])
print(arr[1:4])
print(arr[3:7])
print(arr[:5])
print(arr[3:])
print(arr[::2])
```

```python
# for 2d arry
arr2 = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(arr2[0:2, 1:3])
```

# mathematical operation

```python
arr = np.array([1,2,3,4,5])

print(arr * 2)
print(arr ** 2)
print(arr + 2)

avg = np.mean(arr)
print(arr[arr>avg])
```

```python
arr1 = np.array([2,3,4,5,6])
print(arr + arr1)
print(arr1 - arr)
print(arr/arr1)
print(arr * arr1)
```

```python
# max used numpy functions
marks = np.array([56,89,90,34,50,45])

# average marks
print(np.mean(marks))

# total marks
print(np.sum(marks))

# min marks
print(np.min(marks))

# max marks
print(np.max(marks))

# median marks
print(np.median(marks))

# stadard deviation
print(np.std(marks))
```
