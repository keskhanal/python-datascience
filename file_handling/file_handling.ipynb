# file read

```python
file_path = "student.txt"
file = open(file_path, "r")
print(file.read())
```

file open: open("filepath", "mode")

file modes
- r : read file
- w : write(overwrite)
- a : append
- x : create new file
- rb : read binary
- wb : write binary
- r+ : read & write
- w+ : write & read

```python
# read only 10 chars
filepath = "student.txt"
file = open(filepath, "r")
print(file.read(10))
file.close()
```

```python
# read one line
file = open(filepath, "r")

print(file.readline())
print(file.readline())
print(file.readline())

file.close()
```

```python
# read lines - returns in list
file = open(filepath, "r")
print(file.readlines())
file.close()
```

```python
# using loop
file = open(filepath, "r")
for line in file:
    print(line)
    
file.close()
```

## Write files

```python
# overwrite the file content
filepath = "student.txt"
file = open(filepath, "w")
file.write("Harry")
file.close()
```

```python
# writing multiple lines
filepath = "student.txt"
file = open(filepath, "w")
file.write("Harry\n")
file.write("Ron\n") 
file.write("Hermione\n")
file.close()    
```

## Append mode

```python
file = open(filepath, "a")
file.write("bhagawati\n")
file.close()
```

# create new file

```python
filepath = "fruits.txt"
file = open(filepath, "x")
file.write("apple\n")
```

# closing file

```python
file_path = "student.txt"
file = open(filepath, "r")
file.close()
```

```python
# use with statement to automatically close the file
filepath = "fruits.txt"
with open(filepath, "r") as file:
    print(file.read())
```

```python
# if file already exists, it will throw an error
import os
filepath = "fruits.txt"
if os.path.exists(filepath):
    print("File already exists")
else: 
    print("File does not exist")
```

# delete file

```python
import os
filepath = "fruits.txt"
os.remove(filepath)
```
