```python
# Guess the secret number
sec_num = 7
guess = int(input("Guess the secret number: "))

while guess != sec_num:
  if guess < sec_num:
    print("Guess higher!")

  else:
    print("Guess lower!")

  guess = int(input("Guess the secret number: "))

print("You won!")
```

```python
# atm simulator
balance = 100
pin = 123
steps = input("what you wanna do: ")

if steps == "withdraw":
  amount = float(input("how much: "))
  entered_pin = int(input("enter pin: "))
  if entered_pin == pin:
    if amount<= balance:
      balance -= amount
      print(f"You've withdraw {amount} and your balance is {balance}.")

    else:
      print("insufficient balance")
  else:
    print("invalid pin")

elif steps == "deposit":
  amount = float(input("how much: "))
  entered_pin = int(input("enter pin: "))
  if entered_pin == pin:
    balance += amount
    print(f"You've deposit {amount} and your balance is {balance}")
  else:
    print("invalid pin")

elif steps == "check":
  entered_pin = int(input("enter pin: "))
  if entered_pin == pin:
    print(f"your balance is {balance}.")
  else:
    print("invalid pin!")

elif steps == 'quit':
  print("Exited")

else:
  print("Invalid input")
```

```python
# check prime number
num = int(input("enter a number: "))
for i in range(2, num):
  if num%i == 0:
    print(f"{num} is not a prime number.")
    break

else:
  print(f"{num} is a prime number.")
```

1. reverse word in the sentence ("i love to travel a lot") -> lot a travel to love i
2. check whether given word is pallindrom or not(case insensitive)

```python
# reverse word in sentence
sentence = 'i love to travel a lot'
new_list = sentence.split(" ")

reversed_list = new_list[::-1]
reversed_sen = " ".join(reversed_list)
print(reversed_sen)
```

```python
# check pallindrom (case insensentive)
word = "Rar"
word = word.lower()
reverse_word = word[::-1]
if word == reverse_word:
  print("pallindrom")
else:
  print("not pallindrom")
```
