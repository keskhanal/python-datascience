# Functions revision

```python
# write a function to add two numbers
def add_two_numbers(num1:float, num2:float)->float:
  """add two numbers
  args:
    num1: first number
    num2: second number
  returns:
    sum of two numbers
  """
  result = num1 + num2
  # print(f"sum of two number is: {result}")
  return result
```

```python
num1, num2 = 10, 5
# result = add_two_numbers(num1, num2)
# print(result)
result = add_two_numbers(num1, num2)
print(result)
```
