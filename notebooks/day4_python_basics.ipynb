```python
# can return multiple values
def calculate(num1, num2):
  addition = num1 + num2
  multiplication = num1 * num2
  return addition, multiplication

result1, result2 = calculate(10, 20)
print(result1)
print(result2)
```

### *args vs **kwargs

```python
# use *args when a function may receive an unknown number of postional arguments
def calculate_total(*args):
  total = sum(args)
  return total

print(calculate_total(10, 20))
print(calculate_total(10, 20, 30))
print(calculate_total(10, 20, 30, 40))
```

```python
# use *kwargs when a function may receive an unknown number of keyword arguments
def display_profile(**kwargs):
  for key, value in kwargs.items():
    print(f"{key}: {value}")

display_profile(name="hari", age=25, address="kathmandu")
print("-------------------")
display_profile(name="bikas", gender="male")
```

```python
# function type hints
def calculate_area(length:float, breadth:float) -> float:
  area = length * breadth
  return area

area = calculate_area(10.0, 20.3)
print(type(area))
print(area)
```

```python
# docstring
def calculate_discount(price: float, discount_amt:float) -> float:
  """calculate the discount and return the final price after discount
  args:
    price: float
    discount_amt: float

  returns: float
  """
  final_price = price - discount_amt
  return final_price
```

```python
calculate_discount(price=100, discount_amt=20)
```

```python
# without using max
def largest_number(numbers):
  max_number = float("-inf")

  for num in numbers:
    print(f"max number: {max_number} num: {num}")
    if max_number < num:
      max_number = num

  return max_number


numbers = [-1, -10, 2, 3, 20]
largest_number(numbers)
```

# Exception Handling
- exception is an error that occurs while program is running
- without exception handling, the program may stop immediately

```python
def greeting(name):
  return f"Hi {name}, welcome to python class."

greeting()
```

```python
# try-except
try:
  greeting()

except TypeError as error:
  print(error)
```

```python
try:
  number = int(input("enter any number: "))
  result = 10/number
  print(result)

except ZeroDivisionError:
  print("cannot divide by zero")

except ValueError:
  print("invalid input")

finally:
  print("this is finally block")
```

finally is commonly used for cleanup work such as
- closing a database connection
- closing file
- releasing the system resources
