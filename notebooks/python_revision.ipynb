# *Args vs **Kwargs

```python
# write python functions to add numbers using args
# add_numbers(1, 2, 3) should return 6

# without using args
def add_numbers_without_args(a, b, c):
    return a + b + c

add_numbers_without_args(1, 2, 3)
```

```python
# with using args
def add_numbers_with_args(*args):
    print(args)
    return sum(args)

sum1 = add_numbers_with_args(1, 2, 3, 4, 5)
sum2 = add_numbers_with_args(10, 20, 30)
print(f"Sum1: {sum1}, Sum2: {sum2}")
```

```python
# create a function to print student details using kwargs(name, age, grade)
def student_details(**kwargs):
    print(kwargs)
    # for key, value in kwargs.items():
    #     print(f"{key}: {value}")

student_details(name="John", age=20, grade="A")
```

*args - tuple (a, b)

**kwargs - dictionary {key:value}
