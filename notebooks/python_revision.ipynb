# *Args vs **Kwargs

```python
# write python functions to add numbers using args
# add_numbers(1, 2, 3) should return 6

# without using args
def add_numbers_without_args(a, b, c):
    return a + b + c

add_numbers_without_args(1, 2, 3)
```

```python
# with using args
def add_numbers_with_args(*args):
    print(args)
    return sum(args)

sum1 = add_numbers_with_args(1, 2, 3, 4, 5)
sum2 = add_numbers_with_args(10, 20, 30)
print(f"Sum1: {sum1}, Sum2: {sum2}")
```

```python
# create a function to print student details using kwargs(name, age, grade)
def student_details(**kwargs):
    print(kwargs)
    # for key, value in kwargs.items():
    #     print(f"{key}: {value}")

student_details(name="John", age=20, grade="A")
```

*args - tuple (a, b)

**kwargs - dictionary {key:value}

# Lambda function

```python
# normal function 
def square(num):
    return num ** 2

print(square(5))
```

```python
# using lambda function to square a number
square_lambda = lambda num: num ** 2
print(square_lambda(5))
```

lambda function is a small anonymous functions, especially used when passing functions to other operations.

```python
# example
students = [
    {"name": "John", "age": 20, "grade": "A"},
    {"name": "Jane", "age": 22, "grade": "B"},
    {"name": "Jim", "age": 21, "grade": "C"}
]

students_sorted = sorted(students, key=lambda x: x["age"])
students_sorted
```

# list comprehension

```python
# normal approach
numbers = [1, 2, 3, 4, 5]
squared_numbers = []

for num in numbers:
    squared_numbers.append(num ** 2)

squared_numbers
```

```python
# using list comprehension
squared_numbers_lc = [num ** 2 for num in numbers]
squared_numbers_lc
```

```python
# print even numbers from 1-10 using list comprehension
even_numbers = [num for num in range(1, 11) if num%2 == 0]
even_numbers
```

```python
names = ["John", "Jane", "Jim", "Jack", "Jill", "Harry"]
# using list comprehension get all names in uppercase
uppercase_names = [name.upper() for name in names]
print(uppercase_names)

# name starts with H
names_starting_with_H = [name for name in names if name.startswith("H")]
print(names_starting_with_H)
```

# iterator vs generator

```python
# iterator - for loop handle this process
numbers = [10,20,30,5]
iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
```

```python
# generator
def count_upto(limit):
    number = 1
    while number <= limit:
        yield number
        number += 1

for num in count_upto(10):
    print(num)
```

generator - produces values lazily instead of storing everything in memory at once, make them useful for large dataset.

# decorators

```python
def logger(func):
    def wrapper():
        print("function started.")

        func()

        print("function finished.")

    return wrapper

@logger
def greet():
    print(f"hi, welcome to ds class.")

greet()
```

```python
@logger
def say_hello():
    print("hello")

say_hello()
```

if original function is gift then the decorator is the wrapping paper.
