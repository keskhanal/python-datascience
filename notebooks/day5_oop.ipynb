```python
# functions to check voting eligiblilty based on age and nationality
def is_eligible(age:int, is_nepali:bool=True) -> bool:
  """function to check whether he/she can vote or not
  args:
    age: int
    is_nepali: bool

  returns:
    str

  raise:
    str
  """
  if not is_nepali:
    raise ValueError("Must be nepali citizen to vote.")

  if age<=18:
    raise ValueError("Is too young to vote in this election.")

  return "Eligible to vote."
```

```python
age = 19
is_nepali = True
try:
  result = is_eligible(age, is_nepali)
  print(result)
except Exception as e:
  print(e)
```

# OOP in python
OOP -> object Oriented Programming
- organize code into logical units called classes
- resuing code through inheritance, reducing duplication
- making appplications easier to maintain

```python
def display_student(name, faculty):
  print(f"Student name: {name}")
  print(f"Student faculty: {faculty}")
```

```python
st1_name = "Ram"
st1_fal = "Science"

display_student(st1_name, st1_fal)
```

```python
st2_name = "Ram"
st2_fal = "Science"

display_student(st2_name, st2_fal)
```

```python
# using OOP
class Student:
  def __init__(self, name, faculty):
    self.name = name
    self.faculty = faculty

  def display_student(self):
    print(f"{self.name} is from {self.faculty} faculty.")

  def is_pass(self, marks):
    if marks>40:
      return f"{self.name} is Pass."
    else:
      return f"{self.name} is Fail."
```

```python
student1 = Student("Ram", "Science")
student1.display_student()
check_pass = student1.is_pass(50)
print(check_pass)
```

```python
st2 = Student("Shyam", "Bio")
st2.display_student()
```

- the code is now cleaner, resuable, and much easir to maintain

## Realworld example
Think of a Car:

A Car has:

Attributes(Data):
- Brand
- Name
- color
- model
- speed

Behaviours(Methods):
- start()
- stop()
- Accelerate()
- Brake()


Every individual car(Toyota, Honda, Tesla) is an Object, while the Car is the class(Blueprint) used to create them.

```python
class Car:
  def __init__(self, brand, name, color, model, speed):
    self.brand = brand
    self.name = name
    self.color = color
    self.model = model
    self.speed = speed

  def start(self):
    return f"{self.name} is started."

  def stop(self):
    return f"{self.name} is stopped."

  def accelerate(self):
    print(f"{self.name} is accelerating.")

  def brake(self):
    print(f"{self.name} is braking.")
```

```python
tesla = Car(brand="Tesla", name="cyber truck", color="red", model="2024", speed=200)
print(tesla.model)
print(tesla.stop())
print(tesla.brake())
```

Benefits of OOP:
- better code organization
- code resuability
- easier maintannce
- reduce code duplication
- improved scalabilty
- models real-world problems naturally
