# Pandas Fundamentals
- used for data manipulation and analysis
- numpy is excellent for neumerical arrays, while pandas makes it easier to works with tabular data

```python
import pandas as pd
```

## pandas series

```python
ages = pd.Series([20, 25, 30, 35])

print(ages)
```

```python
# custom index
ages = pd.Series(
    [20, 25, 30, 35], 
    index=['A', 'B', 'C', 'D']
)
print(ages)
```

```python
print(ages["B"])
```

# Dataframe
- two dimensional table containing rows and cols

```python
data = {
    "name": ["Ram", "Sita", "Hari", "Asma", "Bibek", "Sabal"],
    "age": [22, 25, 28, 22, 20, 19],
    "city": ["Kathmandu", "Pokhara", "Chitwan", "Butwal", "Nepalgunj", "Dhading"]
}


data_df = pd.DataFrame(data)
data_df
```
