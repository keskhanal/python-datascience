## Data cleaning

```python
import numpy as np
import pandas as pd

std_data = {
    "name": [
        "Ram", "Sita", "Hari", "Gita", "John", "Rita", "Gita"
    ],
    "age": [20, 24, 22, np.nan, 30, 25, np.nan],
    "city": [
        "Kathmandu", "Pokhara", "Kathmandu", "Chitwan", None, "Kathmandu", "Chitwan"
    ],
    "marks": ["75", "88", "65", "92", "55", "81", "92"]
}

std_df = pd.DataFrame(std_data)
std_df.head(10)
```

```python
# check nan value
std_df.isnull()
```

```python
# count null values
std_df.isnull().sum()
```

```python
std_df.isna().sum()
```

```python
std_df.notna().sum()
```

```python
std_df.isnull().sum().sort_values(ascending=False)
```

```python
# drop missing value
new_df = std_df.dropna()
new_df.head(10)
```

```python
new_df1 = std_df.dropna(subset=["age"])
new_df1
```
