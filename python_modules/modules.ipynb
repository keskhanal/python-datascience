```python
# datetime module
from datetime import datetime
current_datetime = datetime.now()
print(current_datetime)
```

```python
current_date = datetime.today().date()
print(current_date)
```

```python
today = datetime.now()
formatted_date = today.strftime("%d %B %Y")
print(formatted_date)
```

```python
# weekday
weekday = today.strftime("%A")
print(weekday)
```

```python
moth = today.strftime("%B")
print(moth)
```

```python
# age calculator
birthyear = input("Enter your birth year: ")
current_year = datetime.now().year
if current_year < int(birthyear):
    print("You entered a future year. Please enter a valid birth year.")

else:
    age = current_year - int(birthyear)
    print("You are", age, "years old.")
```

# random module

```python
# select random student from a class
students = ["Alice", "Bob", "Charlie", "David", "Eve"]
random_student = random.choice(students)
print("Randomly selected student:", random_student)
```
