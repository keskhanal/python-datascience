## if-elif-else


```
if condition:
  statement
elif condition:
  statement
else:

```



````python
# elif
# check the age group(child, adult, old)
age = 26
if age <= 19:
  print("child")
elif age>19 and age<=45:
  print("adult")
else:
  print("old")
````

## Nested if

### program to check a person if he/she is eligible to vote or not
```
if age greater than 18:
   if nepali citizen:
     can vote
  else
    is not nepali citizen
else
  is too young to vote
```



````python
# program to check a person if he/she is eligible to vote or not
age = 15
nepali_citizen = False

if age >= 18:
  if nepali_citizen:
    print("Can vote.")
  else:
    print("Not a Nepali citizen.")
else:
  print("Too young to vote.")
````

### write a program to check whether given number is odd or even


```
if num % 2 == 0:
  even
else
  odd
```



````python
num = int(input("enter any number: "))
if num % 2 == 0:
  print(f"{num} is a even number.")

else:
  print(f"{num} is a odd number.")
````

# Match Case

````python
day = input("day: ")
match day:
  case "Monday":
    print("Python class.")

  case "Tuesday":
    print("AI class.")

  case "Wednesday":
    print("DS class.")

  case "Thursday":
    print("Python class.")

  case "Friday":
    print("DS class.")

  case _:
    print("No class today.")
````

# While loop


```

```



````python
i = 0
print(f"before: {i}")
i += 1
print(f"After: {i}")
````

````python
num1 = 20
num1 = num1 + 1
num1
````

````python
# suppose you want to print python 100 times
i = 0
print(f"before: {i}")
while i < 100:
  print("Python")
  # i = i + 1
  i += 1
  print(f"After: {i}")
````

# for loop
range(start, stop, step)

````python
for i in range(1, 10, 2):
  print(i)
````

````python
# loop through string
letter = "python"
for char in letter:
  print(char)
````

````python
# break - stops the loop immediately
for i in range(10):
  if i == 5:
    break
  print(i)
````

````python
# continue - skips the current iteration
for i in range(10):
  if i == 5:
    continue
  print(i)
````
