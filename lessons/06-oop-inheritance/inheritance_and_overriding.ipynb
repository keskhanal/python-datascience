### banking system
data(attributes):
  - name
  - address
  - balance
  - pin

behaviour(methods):
- depoist
- withdraw
- check balance
- fund transfer

```python
# banking system
class Bank:
  bank_name = "NIC Asia"

  # constructor (called automatically when object is created)
  def __init__(self, name, address, balance, pin):
    self.name = name
    self.address = address
    self.balance = balance

  def deposit(self, amount):
    if not str(amount).isdigit() or amount <= 0:
      return "invalid amount"

    self.balance += amount
    return self.balance

  def withdraw(self, amount):
    if not str(amount).isdigit() or amount <= 0:
      return "invalid amount"

    if amount > self.balance:
      return "insufficient balance"

    self.balance -= amount
    return self.balance

  def check_balance(self):
    return self.balance
```

# Inheritance
means creating a new class from an existing class

instead of writing code again, we reuse it.

parent class -> child class

```python
# Example

# parent class
class Animal:

  def eat(self):
    print("Animal is eating...")

# child class
class Dog(Animal):
  def bark(self):
    print("Dog is barking...")

class Cat(Animal):
  def meow(self):
    print("Cat is meowing...")

class Tiger(Animal):
  def roar(self):
    print("Tiger is roaring...")
```

```python
d1 = Dog()
print(d1.bark())
print(d1.eat())
```

```python
c1 = Cat()
print(c1.meow())
print(c1.eat())
```

```python
# single inheritance
# Animal -> Dog

# Multiple inheritance
# Animal -> Mammale -> Dog

class Mammale(Animal):
  def walk(self):
    print("Mammale is walking...")

class Dog(Mammale):
  def bark(self):
    print("Dog is barking...")
```

```python
d1 = Dog()
print(d1.bark())
print(d1.walk())
print(d1.eat())

```

```python
# hierarchical inheritance
class Father:
  def skills(self):
    print(f"driving")

class Mother:
  def talent(self):
    print("dancing")

class Child(Father, Mother):
  def play(self):
    print("playing")

  def skills(self):
    print("swimming")
```

```python
c1 = Child()
c1.play()
```

```python
c1.skills()
```

```python
c1.talent()
```

# Method overriding

```python
class Animal:
  def eat(self):
    print("Animal is eating...")

class Dog(Animal):
  def eat(self):
    print("Dog is eating...")
```

```python
d1 = Dog()
d1.eat()
```

```python
# using super()
class Animal:
  def __init__(self, has_legs, lives_in):
    self.has_legs = has_legs
    self.lives_in = lives_in

  def eat(self):
    print("Animal is eating...")

class Dog(Animal):
  def __init__(self, has_legs, lives_in):
    super().__init__(has_legs, lives_in)
    self.bark = "woof woof"

  def eat(self):
    print("Dog is eating...")
```

```python
d1 = Dog(True, "land")
d1.eat()
```

```python
d1.has_legs
```

```python
# magic(dunder) methods
class Student:
  def __init__(self, name, roll_no, marks):
    self.name = name
    self.roll_no = roll_no
    self.marks = marks

  # magic methods
  def __str__(self):
    return f"Name: {self.name}, Roll No: {self.roll_no}, Marks: {self.marks}"

  def __len__(self):
    return 100
```

```python
s1 = Student("suraj", 1, 90)
print(s1)
```

# classwork
1. create Person class and an employee class using inheritance
  - Person
      - attributes(name, age, gender, address)
      - methods(display info)
  - Employee
    - attributes(salary, company(class variable), post, department)
    - methods(display, increase salary, change_dept, annual_salary)

```python
class Person:
  def __init__(self, name, age, gender, address):
    self.name = name
    self.age = age
    self.gender = gender
    self.address = address

  def display_info(self):
    print(f"Name: {self.name}, Age: {self.age}, Gender: {self.gender}, Address: {self.address}")
```

```python
# retired person
p1 = Person(name="john", age=56, gender="male", address="ktm")
p1.display_info()
```

```python
class Employee(Person):
  company_name = "ABC Company"

  def __init__(self, name, age, gender, address, salary, post, department):
    super().__init__(name, age, gender, address)
    self.salary = salary
    self.post = post
    self.department = department

  def display_emp(self):
    super().display_info()
    print(f"Salary: {self.salary}, Post: {self.post}, Department: {self.department}")

  def increase_salary(self, amount:float) -> float:
    self.salary += amount
    return self.salary

  def change_dept(self, new_dept):
    self.department = new_dept
    return self.department

  def annual_salary(self):
    return self.salary * 12
```

```python
e1 = Employee(name="hari", age=26, gender="male", address="ktm", salary=100000, post="manager", department="AI")
e1.display_emp()
```

```python
e1.increase_salary(20000)
```

```python
e1.annual_salary()
```

```python
e1.change_dept("Tech")
```

```python
e1.display_emp()
```
