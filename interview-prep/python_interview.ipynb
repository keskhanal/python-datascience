### write code to find the largest number in list without using max
```text
input: [1, 4, 5, 6, 8] 
output: 8
```

````python
def largest_num(numbers:list)->int:
    """function to return largest number from list

    Args:
        numbers: list
    Returns:
        max_num: int
    """
    max_num = float('-inf')
    for num in numbers:
        if num>max_num:
            max_num = num

    return max_num


numbers = [2,4,5,10,3,20]
largest_num(numbers)
````

### check the char frequency in the given word
```text
input: "banana" 
output: b:1, a:3, n:2
```

````python
def char_count(word:str):
    char_result = {}
    for char in word:
        char_result[char] = char_result.get(char, 0) + 1

    return char_result

word = "banana"
char_count(word)
````

### find the second largest number
```text
input = [10, 20, 5, 30, 25, 30]
output = 25
```

also find the time and space complexity of the code

````python
def second_largest(numbers:list)->int:
    """function to return second largest number from a list
    """
    largest = second_largest = float('-inf')
 
    for num in numbers:
        if num > largest:
            second_largest = largest
            largest = num
 
        elif num > second_largest and num != largest:
            second_largest = num
 
    return second_largest
 
 
numbers = [10, 20, 5, 30, 25, 30]
result = second_largest(numbers)
print(result)
````

Time complexity: O(n)

we loop through the list once
if there are n elements: each element checked once


Space = O(1)

we only use a few extra variables: `largest`, `second`, `n`

the number of variables doesn't increase when the input size increases.
therefore `space = O(1)`
